C1 — Series and DataFrame

```python
import pandas as pd
import numpy as np

# C1(a) Series
marks_series = pd.Series(
    [85, 72, 91, 68, 77],
    index=["Asha", "Bikash", "Sita", "Ram", "Nabin"]
)

print("Series:")
print(marks_series)

print("\nIndex:")
print(marks_series.index)

print("\nValues:")
print(marks_series.values)
```

```python
# C1(b) DataFrame using dictionary of lists

data1 = {
    "student_id": [1, 2, 3, 4, 5],
    "name": ["Asha", "Bikash", "Sita", "Ram", "Nabin"],
    "marks": [85, 72, 91, 68, 77]
}

df1 = pd.DataFrame(data1)

print("DataFrame using dictionary of lists:")
print(df1)
```

```python
# DataFrame using list of dictionaries

data2 = [
    {"student_id": 1, "name": "Asha", "marks": 85},
    {"student_id": 2, "name": "Bikash", "marks": 72},
    {"student_id": 3, "name": "Sita", "marks": 91},
    {"student_id": 4, "name": "Ram", "marks": 68},
    {"student_id": 5, "name": "Nabin", "marks": 77}
]

df2 = pd.DataFrame(data2)

print("DataFrame using list of dictionaries:")
print(df2)

print("\nAre both DataFrames equal?")
print(df1.equals(df2))

print("\nShape:", df1.shape)
print("Columns:", df1.columns)
print("Index:", df1.index)
```

C2 — Read and Inspect CSV

```python
# C2
df = pd.read_csv("../data/scores_raw.csv")

print("First 5 rows:")
print(df.head())

print("\nLast 3 rows:")
print(df.tail(3))

print("\nShape:")
print(df.shape)

print("\nInfo:")
df.info()

print("\nNumeric description:")
print(df.describe())

print("\nObject description:")
print(df.describe(include="object"))

print("""
C2 observations:
1. The dataset contains inconsistent city spellings/capitalization.
2. Some attendance or marks values are missing.
3. There are duplicate rows that need to be removed before analysis.
""")
```

C3 — Selecting Columns and Rows

```python
# Single column
print("df['marks']:")
print(df["marks"])

print("\nType:")
print(type(df["marks"]))

# DataFrame with one column
print("\ndf[['marks']]:")
print(df[["marks"]])

print("\nType:")
print(type(df[["marks"]]))
```

C4 — Filtering

```python
# C4(a)
filter1 = df[
    (df["marks"] > 80) &
    (df["subject"] == "Python Programming")
]

print("Marks > 80 and Python Programming:")
print(filter1)

filter2 = df[
    df["city"].isin(["Kathmandu", "Pokhara"])
]

print("\nCity is Kathmandu or Pokhara:")
print(filter2)
```

```python
# C4(b) Using query

query_result = df.query(
    "marks > 80 and subject == 'Python Programming'"
)

print("Using query:")
print(query_result)

print("\nSame result?")
print(filter1.equals(query_result))

print("""
query() allows filtering using a string expression.
It can make multiple conditions easier to read.
""")
```

C5 — Cleaning City Names

```python
# Before cleaning
print("Number of unique cities before cleaning:")
print(df["city"].nunique())

print("\nCity counts before cleaning:")
print(df["city"].value_counts())
```

```python
# Clean city names
df["city"] = (
    df["city"]
    .astype(str)
    .str.strip()
    .str.capitalize()
)

print("Number of unique cities after cleaning:")
print(df["city"].nunique())

print("\nCity counts after cleaning:")
print(df["city"].value_counts())
```

```python
# Re-run C4(b) after cleaning
clean_city_filter = df[
    df["city"].isin(["Kathmandu", "Pokhara"])
]

print("Kathmandu or Pokhara after cleaning:")
print(clean_city_filter)

print("\nRows found:", len(clean_city_filter))
```

C6 — Missing Values

```python
print("Missing values:")
print(df.isna().sum())
```

```python
# Fill missing attendance with mean
rows_before_attendance = len(df)

attendance_mean = df["attendance_percent"].mean()

df["attendance_percent"] = df["attendance_percent"].fillna(
    attendance_mean
)

rows_after_attendance = len(df)

print("Rows before filling attendance:", rows_before_attendance)
print("Rows after filling attendance:", rows_after_attendance)
print("Attendance mean used:", attendance_mean)
```

```python
# Drop rows with missing marks
rows_before_marks = len(df)

df = df.dropna(subset=["marks"])

rows_after_marks = len(df)

print("Rows before dropping missing marks:", rows_before_marks)
print("Rows after dropping missing marks:", rows_after_marks)

print("""
Using the mean for missing attendance keeps the rows in the dataset,
but the mean value may hide the actual attendance of an individual student.
""")
```

C7 — Duplicate Rows

```python
# Load original data again for the duplicate checkpoint
df_duplicates = pd.read_csv("../data/scores_raw.csv")

duplicate_mask = df_duplicates.duplicated()

print("Number of duplicate rows:")
print(duplicate_mask.sum())

print("\nSome duplicate rows:")
print(df_duplicates[duplicate_mask].head())
```

```python
before_duplicates = df_duplicates.shape

df_duplicates = df_duplicates.drop_duplicates()

after_duplicates = df_duplicates.shape

print("Shape before removing duplicates:", before_duplicates)
print("Shape after removing duplicates:", after_duplicates)

# Expected checkpoint: (126, columns) -> (120, columns)
```

```python
# Duplicates based only on student_id + subject
subset_duplicates = df_duplicates.duplicated(
    subset=["student_id", "subject"]
).sum()

print("Duplicates based on student_id + subject:")
print(subset_duplicates)

print("""
A student_id + subject combination should normally identify one
student's record for that subject. Duplicate combinations can therefore
indicate repeated records even when other columns are different.
""")
```

```python
# Final cleaned dataframe
df_clean = pd.read_csv("../data/scores_raw.csv")

# Clean city
df_clean["city"] = (
    df_clean["city"]
    .astype(str)
    .str.strip()
    .str.capitalize()
)

# Fill missing attendance
df_clean["attendance_percent"] = df_clean["attendance_percent"].fillna(
    df_clean["attendance_percent"].mean()
)

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Remove rows with missing marks
df_clean = df_clean.dropna(subset=["marks"])

print("Final cleaned shape:", df_clean.shape)
```

C8 — Dates

```python
print("Original exam_date dtype:")
print(df_clean["exam_date"].dtype)

df_clean["exam_date"] = pd.to_datetime(
    df_clean["exam_date"]
)

df_clean["exam_month"] = df_clean["exam_date"].dt.month
df_clean["exam_weekday"] = df_clean["exam_date"].dt.day_name()

print("\nUpdated exam_date dtype:")
print(df_clean["exam_date"].dtype)

print("\nDate columns:")
print(df_clean[["exam_date", "exam_month", "exam_weekday"]].head())
```

```python
after_date = df_clean[
    df_clean["exam_date"] > "2026-02-15"
]

print("Rows after 2026-02-15:")
print(after_date)

print("\nCount:", len(after_date))
```

C10 — Pass and Grade

```python
# Passed column
df_clean["passed"] = df_clean["marks"] >= 40

# Grade using nested np.where
df_clean["grade"] = np.where(
    df_clean["marks"] >= 85, "A",
    np.where(
        df_clean["marks"] >= 70, "B",
        np.where(
            df_clean["marks"] >= 40, "C", "F"
        )
    )
)

print(df_clean[["marks", "passed", "grade"]].head(20))
```

```python
pass_rate = df_clean["passed"].mean() * 100

print("Pass rate:", pass_rate, "%")

print("\nGrade counts:")
print(df_clean["grade"].value_counts())
```

```python
# Check impossible marks
impossible_marks = df_clean[
    (df_clean["marks"] < 0) |
    (df_clean["marks"] > 100)
]

print("Impossible marks:")
print(impossible_marks)

print("""
Marks above 100 are impossible if marks are recorded out of 100.
Such values should be investigated and corrected or treated as invalid
before using the data for final analysis.
""")
```

C11 — Category Data Type

```python
# Memory before
memory_before = df_clean[["subject", "city"]].memory_usage(
    deep=True
).sum()

print("Memory before category conversion:", memory_before, "bytes")

# Convert to category
df_clean["subject"] = df_clean["subject"].astype("category")
df_clean["city"] = df_clean["city"].astype("category")

# Memory after
memory_after = df_clean[["subject", "city"]].memory_usage(
    deep=True
).sum()

print("Memory after category conversion:", memory_after, "bytes")

memory_saved = memory_before - memory_after
percentage_saved = (memory_saved / memory_before) * 100

print("Memory saved:", memory_saved, "bytes")
print("Percentage saved:", percentage_saved, "%")

print("""
Categorical data is useful when a column contains a limited number
of repeated values, because pandas stores category codes instead of
repeating the full strings. It may provide less benefit when most
values are unique.
""")
```

C12 — GroupBy and Pivot Tables

```python
# C12(a)
subject_summary = (
    df_clean
    .groupby("subject", observed=True)["marks"]
    .agg(["mean", "max", "count"])
)

print("Subject summary:")
print(subject_summary)
```

```python
# C12(b)
city_subject_mean = (
    df_clean
    .groupby(["city", "subject"], observed=True)["marks"]
    .mean()
)

print("Mean marks by city and subject:")
print(city_subject_mean)
```

```python
# C12(c)
pivot_table = pd.pivot_table(
    df_clean,
    values="marks",
    index="city",
    columns="subject",
    aggfunc="mean"
)

print("Pivot table:")
print(pivot_table)
```
