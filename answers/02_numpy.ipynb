```python
import numpy as np

np.random.seed(7)

cities = np.array([
    "Kathmandu",
    "Pokhara",
    "Biratnagar",
    "Lalitpur",
    "Birgunj"
])

days = np.array([f"D{i:02d}" for i in range(1, 15)])

temps = np.random.randint(14, 36, size=(5, 14)).astype(float)
humidity = np.random.randint(40, 95, size=(5, 14)).astype(float)

temps[2, 5] = np.nan
temps[4, 11] = np.nan
```

B1 — Array anatomy and reshaping

```python
print("\n--- B1 ---")

print("B1 — shape:", temps.shape)
print("B1 — ndim:", temps.ndim)
print("B1 — dtype:", temps.dtype)
print("B1 — size:", temps.size)

arr = np.arange(70)

reshaped_explicit = arr.reshape(5, 14)
reshaped_auto = arr.reshape(5, -1)

print("B1 — reshape with explicit dimensions:")
print(reshaped_explicit)

print("B1 — reshape using -1:")
print(reshaped_auto)

print("B1 — reshaped arrays identical:",
      np.array_equal(reshaped_explicit, reshaped_auto))

flattened = reshaped_explicit.flatten()

print("B1 — flattened:")
print(flattened)

# -1 tells NumPy to automatically calculate the missing dimension.
# Reshaping normally creates a different view of the same data when possible,
# rather than copying the underlying data.
```

B2 — Indexing and slicing

```python
print("\n--- B2 ---")

# (a) Pokhara's day-3 temperature
result = temps[1, 2]
print("B2(a) — Pokhara day-3 temperature:", result)
print("B2(a) — shape:", np.shape(result))

# (b) Every city's reading on the last day
result = temps[:, -1]
print("B2(b) — last day readings:", result)
print("B2(b) — shape:", result.shape)

# (c) Biratnagar's full fortnight
result = temps[2, :]
print("B2(c) — Biratnagar fortnight:", result)
print("B2(c) — shape:", result.shape)

# (d) First three cities across days 4–8
result = temps[:3, 3:8]
print("B2(d) — first 3 cities, days 4–8:")
print(result)
print("B2(d) — shape:", result.shape)

# (e) Every other day for Lalitpur
result = temps[3, ::2]
print("B2(e) — Lalitpur every other day:", result)
print("B2(e) — shape:", result.shape)

# (f) Kathmandu's fortnight reversed
result = temps[0, ::-1]
print("B2(f) — Kathmandu reversed:", result)
print("B2(f) — shape:", result.shape)
```

B3 — Boolean masks

```python
print("\n--- B3 ---")

hot_mask = temps > 30

print("B3 — hot mask:")
print(hot_mask)

hot_count_per_city = np.sum(hot_mask, axis=1)

print("B3 — readings above 30 per city:")
print(hot_count_per_city)

kathmandu_hot_days = days[hot_mask[0]]

print("B3 — Kathmandu days above 30:")
print(kathmandu_hot_days)

# A mask is a Boolean array containing True/False values.
# temps[temps > 30] returns matching values as a 1D result,
# so it does not preserve the original 2D shape.
```

B4 — Views vs copies


```python
print("\n--- B4 ---")

# Work on a copy of temps so our original dataset is preserved
view_test = temps.copy()

city_slice = view_test[0, :3]

print("B4 — original first 3 Kathmandu values:")
print(view_test[0, :3])

city_slice[0] = 999

print("B4 — after changing the slice:")
print(view_test[0, :3])

# Now demonstrate copy()
copy_test = temps.copy()

city_copy = copy_test[0, :3].copy()

print("B4 — original before changing copied slice:")
print(copy_test[0, :3])

city_copy[0] = 888

print("B4 — original after changing copied slice:")
print(copy_test[0, :3])

# A slice usually creates a view that refers to the same underlying data.
# Forgetting this can cause an accidental change to the original array.
# .copy() creates independent data, so changing it does not affect the original.
```

B5 — Vectorization vs a Python loop

```python

print("\n--- B5 ---")

# Small example
small_temps = np.array([20.0, 25.0, 30.0])

# Explicit Python loop
fahrenheit_loop = []
for temp in small_temps:
    fahrenheit_loop.append((temp * 9 / 5) + 32)

fahrenheit_loop = np.array(fahrenheit_loop)

# Vectorized operation
fahrenheit_vectorized = (small_temps * 9 / 5) + 32

print("B5 — loop result:", fahrenheit_loop)
print("B5 — vectorized result:", fahrenheit_vectorized)
print("B5 — results agree:",
      np.allclose(fahrenheit_loop, fahrenheit_vectorized))

# Timing on 1,000,000 values
large_array = np.random.uniform(14, 36, 1_000_000)

start = __import__("time").perf_counter()

loop_result = []
for temp in large_array:
    loop_result.append((temp * 9 / 5) + 32)

loop_result = np.array(loop_result)

loop_time = __import__("time").perf_counter() - start

start = __import__("time").perf_counter()

vector_result = (large_array * 9 / 5) + 32

vector_time = __import__("time").perf_counter() - start

speedup = loop_time / vector_time

print("B5 — loop timing:", loop_time, "seconds")
print("B5 — vectorized timing:", vector_time, "seconds")
print("B5 — speedup:", speedup, "x faster")

# Python loops process values one at a time.
# NumPy performs vectorized operations using optimized array operations,
# which avoids much of the Python-level loop overhead.
```

B6 — Broadcasting

```python

print("\n--- B6 ---")

# (a) Average temperature for each city
city_average = np.nanmean(temps, axis=1, keepdims=True)

anomalies = temps - city_average

print("B6(a) — city averages shape:", city_average.shape)
print("B6(a) — anomaly grid:")
print(anomalies)
print("B6(a) — anomaly shape:", anomalies.shape)

# (b) Per-day sensor correction
daily_correction = np.arange(14) * 0.1

corrected_temps = temps + daily_correction

print("B6(b) — daily correction shape:", daily_correction.shape)
print("B6(b) — corrected temperatures:")
print(corrected_temps)

# (c) Intentional broadcasting error
try:
    result = temps + np.array([1, 2, 3, 4, 5])
except ValueError as e:
    print("B6(c) — broadcasting error:", e)

# Broadcasting compares shapes from the right.
# (5, 14) and (14,) can align because 14 matches the last dimension.
# (5, 14) and (5,) cannot align because 5 does not match 14.
```

B7 — Matrix multiplication

```python

print("\n--- B7 ---")

quantities = np.array([2, 3, 1])
prices = np.array([120, 90, 250])

print("B7(a) — quantities * prices:",
      quantities * prices)

print("B7(a) — quantities @ prices:",
      quantities @ prices)

# * performs element-by-element multiplication.
# @ performs matrix/vector multiplication and produces the dot product here.

# Features: 14 days × 2 measurements
mean_temperature = np.nanmean(temps, axis=0)
mean_humidity = np.nanmean(humidity, axis=0)

features = np.column_stack([
    mean_temperature,
    mean_humidity
])

weights = np.array([0.6, -0.1])

comfort_score = features @ weights

print("B7(b) — features shape:", features.shape)
print("B7(b) — weights shape:", weights.shape)
print("B7(b) — comfort score shape:", comfort_score.shape)
print("B7(b) — comfort scores:", comfort_score)

# The inner dimensions must match:
# (14, 2) @ (2,) produces (14,).
```

B8 — Aggregating along an axis

```python

print("\n--- B8 ---")

city_mean = np.nanmean(temps, axis=1)
city_max = np.nanmax(temps, axis=1)
city_min = np.nanmin(temps, axis=1)
city_std = np.nanstd(temps, axis=1)

print("B8 — mean per city:", city_mean)
print("B8 — mean shape:", city_mean.shape)

print("B8 — max per city:", city_max)
print("B8 — max shape:", city_max.shape)

print("B8 — min per city:", city_min)
print("B8 — min shape:", city_min.shape)

print("B8 — std per city:", city_std)
print("B8 — std shape:", city_std.shape)

day_mean = np.nanmean(temps, axis=0)
day_max = np.nanmax(temps, axis=0)

print("B8 — mean per day:", day_mean)
print("B8 — mean shape:", day_mean.shape)

print("B8 — max per day:", day_max)
print("B8 — max shape:", day_max.shape)

hottest_day_indices = np.nanargmax(temps, axis=1)
hottest_days = days[hottest_day_indices]

print("B8 — hottest day for each city:", hottest_days)
print("B8 — hottest days shape:", hottest_days.shape)

# axis=0 collapses the city dimension, leaving one value per day.
# axis=1 collapses the day dimension, leaving one value per city.
```

B9 — Reproducible randomness and train/test split

```python

print("\n--- B9 ---")

flat_temps = temps.flatten()

np.random.seed(42)

indices = np.arange(flat_temps.size)
np.random.shuffle(indices)

split_point = int(0.8 * len(indices))

train_indices = indices[:split_point]
test_indices = indices[split_point:]

print("B9 — train shape:", train_indices.shape)
print("B9 — test shape:", test_indices.shape)
print("B9 — total samples:", len(train_indices) + len(test_indices))
print("B9 — no shared indices:",
      len(np.intersect1d(train_indices, test_indices)) == 0)

# Repeat with the same seed
np.random.seed(42)

indices_again = np.arange(flat_temps.size)
np.random.shuffle(indices_again)

train_again = indices_again[:split_point]
test_again = indices_again[split_point:]

print("B9 — train split reproducible:",
      np.array_equal(train_indices, train_again))

print("B9 — test split reproducible:",
      np.array_equal(test_indices, test_again))

# Shuffling prevents the split from depending on the original ordering.
# The seed makes the random split reproducible when results are shared.
```

B10 — np.where, argmax, argmin

```python

print("\n--- B10 ---")

# (a) Temperature labels
labels_grid = np.where(
    temps > 30,
    "hot",
    np.where(
        temps >= 22,
        "mild",
        "cold"
    )
)

print("B10(a) — temperature label grid:")
print(labels_grid)

print("B10(a) — label grid shape:", labels_grid.shape)

# (b) Predicted class
scores = np.array([0.08, 0.62, 0.21, 0.09])
labels = np.array(["rain", "clear", "cloudy", "fog"])

predicted_index = np.argmax(scores)
predicted_label = labels[predicted_index]

print("B10(b) — predicted label:", predicted_label)

# argmax gives the position of the largest score,
# which lets us use that position to select the corresponding class label.
# max() would only give the score, not the class name.
```
